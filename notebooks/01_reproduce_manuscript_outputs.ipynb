## 1. Setup

This step locates the repository root and defines the main folders used by the notebook.

```python
from pathlib import Path
import pandas as pd
import numpy as np
from IPython.display import display, Markdown, Image

candidate_roots = [
    Path.cwd(),
    Path.cwd().parent,
    Path("/content/drive/MyDrive/repository_exports/xai-coupled-nmf-recommendation"),
]

REPO_ROOT = None

for root in candidate_roots:
    if (root / "results").exists() and (root / "paper_assets").exists():
        REPO_ROOT = root
        break

if REPO_ROOT is None:
    raise FileNotFoundError(
        "Could not locate repository root. "
        "If running in Colab, mount Google Drive and check the repository path."
    )

CSV_DIR = REPO_ROOT / "results" / "csv"
TABLE_DIR = REPO_ROOT / "paper_assets" / "tables"
FIGURE_DIR = REPO_ROOT / "paper_assets" / "figures"
DOCS_DIR = REPO_ROOT / "docs"
CONFIG_DIR = REPO_ROOT / "configs"

print("REPO_ROOT:", REPO_ROOT)
print("CSV_DIR:", CSV_DIR)
print("TABLE_DIR:", TABLE_DIR)
print("FIGURE_DIR:", FIGURE_DIR)
print("DOCS_DIR:", DOCS_DIR)
print("CONFIG_DIR:", CONFIG_DIR)
```

## 2. Verify Key Repository Files

This step checks whether the main repository documentation and configuration files are present.

```python
required_paths = [
    REPO_ROOT / "README.md",
    REPO_ROOT / "requirements.txt",
    REPO_ROOT / "environment.yml",
    CONFIG_DIR / "ml_latest_small_config.yaml",
    CONFIG_DIR / "ml1m_config.yaml",
    CONFIG_DIR / "selected_hyperparameters.yaml",
    DOCS_DIR / "reproduction_guide.md",
    DOCS_DIR / "table_figure_mapping.md",
    DOCS_DIR / "experiment_summary.md",
]

verify_rows = []

for path in required_paths:
    verify_rows.append({
        "file": str(path.relative_to(REPO_ROOT)),
        "exists": path.exists(),
        "size_kb": round(path.stat().st_size / 1024, 2) if path.exists() else None,
    })

verify_df = pd.DataFrame(verify_rows)
display(verify_df)

missing = verify_df[verify_df["exists"] == False]

if len(missing) == 0:
    print("All key repository files are present.")
else:
    print("Missing key repository files:")
    display(missing)
```

## 7. Preview Manuscript Figures

This step displays the PNG version of each manuscript figure inside the notebook.

The preview is only for visual inspection. The manuscript should use the corresponding `.pdf` files in LaTeX/Overleaf.

If a PNG preview is missing but the PDF exists, the figure is still available for manuscript compilation.

```python
for item in expected_figures:
    role = item["role"]
    png_path = FIGURE_DIR / item["png"]
    pdf_path = FIGURE_DIR / item["pdf"]

    display(Markdown(f"### {role}"))

    if png_path.exists():
        display(Image(filename=str(png_path)))
    elif pdf_path.exists():
        print("PNG preview not available, but the PDF figure exists for LaTeX use:")
        print(pdf_path)
    else:
        print("Figure file is missing:")
        print(item["pdf"])
```

## 9. Main Manuscript Interpretation

The final experimental evidence supports the following interpretation:

1. On MovieLens latest-small, coupled NMF remains statistically comparable to plain NMF while providing descriptor-grounded additive explanations.

2. Biased MF remains stronger for rating prediction metrics such as RMSE and MAE.

3. The explanation fidelity results show that recommendation explanations are distributed across multiple latent factors.

4. On MovieLens 1M, tuned coupled NMF is weaker than plain NMF on RMSE and MAE but remains competitive on top-$K$ ranking metrics.

5. Descriptor richness matters: tag-and-genre descriptors provide stronger semantic grounding than genre-only descriptors.
